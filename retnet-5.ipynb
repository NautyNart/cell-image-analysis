## ResNet-50 Model

```python
import os, random, time, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image
from sklearn.metrics import (f1_score, accuracy_score, cohen_kappa_score,
                             confusion_matrix, classification_report)
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
warnings.filterwarnings('ignore')

# CONFIG
PREPROCESSED_DIR = '/kaggle/input/datasets/eva233333/data3888-preprocessing-output/'
OUT_DIR          = '/kaggle/working/'

SEED         = 42
NUM_EPOCHS   = 30
PATIENCE     = 5
BATCH_SIZE   = 64
NUM_CLASSES  = 4
IMG_SIZE_100 = 224
IMG_SIZE_50  = 140
DEVICE       = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

LR_HEAD   = 3e-4
LR_LAYER4 = 3e-5

CLASS_NAMES = ['DCIS_1', 'DCIS_2', 'Prolif_Invasive_Tumor', 'Invasive_Tumor']

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)


# LOAD DATASETS
master_s2  = pd.read_csv(PREPROCESSED_DIR + 'master_labels_s2.csv')
norm_stats = pd.read_csv(PREPROCESSED_DIR + 'norm_stats.csv')
MEAN       = norm_stats['mean'].tolist()
STD        = norm_stats['std'].tolist()

print("Strategy 2 (50px / 100px CNN)")
print(f"Total: {len(master_s2):,}  |  Augmented: {master_s2['augmented'].sum():,}")
print(master_s2['split'].value_counts().to_string())
print("\nClass distribution:")
print(master_s2['cluster'].value_counts().to_string())


# DATASET
MASK_FILL       = np.array([123, 116, 103], dtype=np.uint8)
MASK_KEEP_RATIO = 0.30

def apply_mask(img):
    arr    = np.array(img)
    h, w   = arr.shape[:2]
    cx, cy = w // 2, h // 2
    radius = int(min(h, w) * MASK_KEEP_RATIO)
    Y, X   = np.ogrid[:h, :w]
    inside = (X - cx) ** 2 + (Y - cy) ** 2 <= radius ** 2
    result = np.full_like(arr, fill_value=MASK_FILL)
    result[inside] = arr[inside]
    return Image.fromarray(result)

def get_transforms(use_100px=True):
    img_size = IMG_SIZE_100 if use_100px else IMG_SIZE_50
    return transforms.Compose([
        transforms.Resize((img_size, img_size)),
        transforms.ToTensor(),
        transforms.Normalize(mean=MEAN, std=STD),
    ])


class CellDataset(Dataset):
    def __init__(self, df, split, use_100px=True, use_masked=False):
        self.df         = df[df['split'] == split].reset_index(drop=True)
        self.transform  = get_transforms(use_100px)
        self.use_masked = use_masked
        self.path_col   = 'path_100' if use_100px else 'path_50'

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        img = Image.open(row[self.path_col]).convert('RGB')

        # Apply rotation for augmented Prolif images
        if row['augmented'] and row['rotation'] != 0:
            img = img.rotate(row['rotation'])

        if self.use_masked:
            img = apply_mask(img)

        img   = self.transform(img)
        label = torch.tensor(row['label'], dtype=torch.long)
        return img, label


def get_dataloaders(master_df, use_100px=True, use_masked=False):
    train_ds = CellDataset(master_df, 'train', use_100px, use_masked)
    val_ds   = CellDataset(master_df, 'val',   use_100px, use_masked)
    test_ds  = CellDataset(master_df, 'test',  use_100px, use_masked)
    train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True,
                              num_workers=2, pin_memory=True)
    val_loader   = DataLoader(val_ds,   batch_size=BATCH_SIZE, shuffle=False,
                              num_workers=2, pin_memory=True)
    test_loader  = DataLoader(test_ds,  batch_size=BATCH_SIZE, shuffle=False,
                              num_workers=2, pin_memory=True)
    return train_loader, val_loader, test_loader


# Verify dataset sizes
print("\nStrategy 2 dataset sizes")
for px, flag in [('100px', True), ('50px', False)]:
    tr = CellDataset(master_s2, 'train', flag)
    va = CellDataset(master_s2, 'val',   flag)
    te = CellDataset(master_s2, 'test',  flag)
    print(f"  {px}: Train {len(tr):,} | Val {len(va):,} | Test {len(te):,}")

print("\nStrategy 3 dataset sizes (masked)")
tr_m = CellDataset(master_s2, 'train', use_100px=True, use_masked=True)
va_m = CellDataset(master_s2, 'val',   use_100px=True, use_masked=True)
te_m = CellDataset(master_s2, 'test',  use_100px=True, use_masked=True)
print(f"  masked 100px: Train {len(tr_m):,} | Val {len(va_m):,} | Test {len(te_m):,}")

# Quick batch shape check
sample_loader = DataLoader(CellDataset(master_s2, 'train', True),  batch_size=4)
imgs, _       = next(iter(sample_loader))
print(f"\n  100px batch shape:    {imgs.shape}")   # [4, 3, 224, 224]

sample_loader_50 = DataLoader(CellDataset(master_s2, 'train', False), batch_size=4)
imgs_50, _       = next(iter(sample_loader_50))
print(f"  50px batch shape:     {imgs_50.shape}")  # [4, 3, 140, 140]

sample_loader_m = DataLoader(CellDataset(master_s2, 'train', True, True), batch_size=4)
imgs_m, _       = next(iter(sample_loader_m))
print(f"  masked 100px shape:   {imgs_m.shape}")   # [4, 3, 224, 224]


# MODEL — ResNet-50
def build_model():
    model    = models.resnet50(weights='IMAGENET1K_V2')
    model.fc = nn.Sequential(
        nn.Dropout(p=0.4),
        nn.Linear(2048, NUM_CLASSES)
    )
    return model.to(DEVICE)


def get_optimizer(model):
    for param in model.parameters():
        param.requires_grad = False
    for param in model.layer4.parameters():
        param.requires_grad = True
    for param in model.fc.parameters():
        param.requires_grad = True
    return torch.optim.Adam([
        {'params': model.layer4.parameters(), 'lr': LR_LAYER4},
        {'params': model.fc.parameters(),     'lr': LR_HEAD},
    ], weight_decay=1e-3)


# TRAINING AND EVALUATION
def train_one_epoch(model, loader, optimizer, criterion):
    model.train()
    total_loss, all_preds, all_labels = 0.0, [], []
    for imgs, labels in loader:
        imgs, labels = imgs.to(DEVICE), labels.to(DEVICE)
        optimizer.zero_grad()
        outputs = model(imgs)
        loss    = criterion(outputs, labels)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()
        total_loss += loss.item() * imgs.size(0)
        preds = outputs.argmax(dim=1)
        all_preds.extend(preds.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())
    avg_loss = total_loss / len(loader.dataset)
    acc      = accuracy_score(all_labels, all_preds)
    f1       = f1_score(all_labels, all_preds, average='macro', zero_division=0)
    return avg_loss, acc, f1


def evaluate(model, loader, criterion):
    model.eval()
    total_loss, all_preds, all_labels = 0.0, [], []
    with torch.no_grad():
        for imgs, labels in loader:
            imgs, labels = imgs.to(DEVICE), labels.to(DEVICE)
            outputs = model(imgs)
            loss    = criterion(outputs, labels)
            total_loss += loss.item() * imgs.size(0)
            preds = outputs.argmax(dim=1)
            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())
    avg_loss = total_loss / len(loader.dataset)
    acc      = accuracy_score(all_labels, all_preds)
    f1       = f1_score(all_labels, all_preds, average='macro', zero_division=0)
    return avg_loss, acc, f1, all_preds, all_labels


# MAIN TRAINING LOOP
def train_cnn(master_df, use_100px=True, use_masked=False, tag=None):
    if tag is None:
        tag = ('masked' if use_masked else ('100px' if use_100px else '50px'))

    print(f"Training CNN — {tag}")

    train_loader, val_loader, test_loader = get_dataloaders(
        master_df, use_100px, use_masked)
    print(f"Train: {len(train_loader.dataset):,} | "
          f"Val:   {len(val_loader.dataset):,} | "
          f"Test:  {len(test_loader.dataset):,}")

    model     = build_model()
    optimizer = get_optimizer(model)
    criterion = nn.CrossEntropyLoss()
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer, mode='max', factor=0.5, patience=3)

    history = {'train_loss': [], 'val_loss': [],
               'train_acc':  [], 'val_acc':  []}

    best_val_f1     = 0.0
    patience_count  = 0
    best_model_path = OUT_DIR + f'cnn_best_{tag}.pth'

    for epoch in range(1, NUM_EPOCHS + 1):
        t0 = time.time()
        train_loss, train_acc, train_f1 = train_one_epoch(
            model, train_loader, optimizer, criterion)
        val_loss, val_acc, val_f1, _, _ = evaluate(
            model, val_loader, criterion)
        scheduler.step(val_f1)

        history['train_loss'].append(train_loss)
        history['val_loss'].append(val_loss)
        history['train_acc'].append(train_acc)
        history['val_acc'].append(val_acc)

        print(f"Epoch {epoch:02d}/{NUM_EPOCHS} | "
              f"Train Loss: {train_loss:.4f} | Train Acc: {train_acc:.4f} | "
              f"Val Loss: {val_loss:.4f} | Val Acc: {val_acc:.4f} | "
              f"Time: {time.time()-t0:.1f}s")

        if val_f1 > best_val_f1:
            best_val_f1    = val_f1
            patience_count = 0
            torch.save(model.state_dict(), best_model_path)
            print(f"   Best model saved (val F1: {best_val_f1:.4f})")
        else:
            patience_count += 1
            if patience_count >= PATIENCE:
                print(f"  Early stopping at epoch {epoch}")
                break

    # Test evaluation
    model.load_state_dict(torch.load(best_model_path))
    test_loss, test_acc, test_f1, test_preds, test_labels = evaluate(
        model, test_loader, criterion)
    kappa = cohen_kappa_score(test_labels, test_preds)

    print(f"\nTEST RESULTS ({tag})")
    print(f"  Test Loss:     {test_loss:.4f}")
    print(f"  Test Accuracy: {test_acc:.4f}")
    print(f"  Test Macro F1: {test_f1:.4f}")
    print(f"  Cohen's Kappa: {kappa:.4f}")
    print(f"\nClassification Report:")
    print(classification_report(test_labels, test_preds,
                                target_names=CLASS_NAMES,
                                zero_division=0))

    # Training curves
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    fig.suptitle(f'CNN Training History — {tag}', fontweight='bold')
    axes[0].plot(history['train_loss'], label='Train', color='#457B9D')
    axes[0].plot(history['val_loss'],   label='Val',   color='#E63946')
    axes[0].set_xlabel('Epoch'); axes[0].set_ylabel('Loss')
    axes[0].set_title('Loss');   axes[0].legend()
    axes[1].plot(history['train_acc'], label='Train', color='#457B9D')
    axes[1].plot(history['val_acc'],   label='Val',   color='#E63946')
    axes[1].set_xlabel('Epoch'); axes[1].set_ylabel('Accuracy')
    axes[1].set_title('Accuracy'); axes[1].legend()
    plt.tight_layout()
    plt.savefig(OUT_DIR + f'cnn_history_{tag}.png', dpi=150, bbox_inches='tight')
    plt.show()

    # Confusion matrix
    cm = confusion_matrix(test_labels, test_preds)
    fig, ax = plt.subplots(figsize=(7, 6))
    im = ax.imshow(cm, cmap='Blues')
    plt.colorbar(im)
    ax.set_xticks(range(NUM_CLASSES)); ax.set_yticks(range(NUM_CLASSES))
    ax.set_xticklabels(CLASS_NAMES, rotation=30, ha='right', fontsize=9)
    ax.set_yticklabels(CLASS_NAMES, fontsize=9)
    ax.set_xlabel('Predicted'); ax.set_ylabel('True')
    ax.set_title(f'Confusion Matrix — CNN {tag}', fontweight='bold')
    for i in range(NUM_CLASSES):
        for j in range(NUM_CLASSES):
            ax.text(j, i, cm[i, j], ha='center', va='center',
                    color='white' if cm[i, j] > cm.max() / 2 else 'black',
                    fontsize=9)
    plt.tight_layout()
    plt.savefig(OUT_DIR + f'cnn_confusion_{tag}.png', dpi=150, bbox_inches='tight')
    plt.show()

    # Save results
    results = {
        'tag':           tag,
        'test_loss':     test_loss,
        'test_macro_f1': test_f1,
        'test_accuracy': test_acc,
        'cohen_kappa':   kappa,
    }
    pd.DataFrame([results]).to_csv(
        OUT_DIR + f'cnn_results_{tag}.csv', index=False)

    return model, history, results


# RUN — three experiments
model_100,    history_100,    results_100    = train_cnn(master_s2, use_100px=True,  use_masked=False, tag='100px')
model_50,     history_50,     results_50     = train_cnn(master_s2, use_100px=False, use_masked=False, tag='50px')
model_masked, history_masked, results_masked = train_cnn(master_s2, use_100px=True,  use_masked=True,  tag='masked')

# Final comparison
print("FINAL COMPARISON: 50px vs 100px vs masked")
print(f"{'Metric':<20} {'100px':>10} {'50px':>10} {'masked':>10}")
for key in ['test_macro_f1', 'test_accuracy', 'test_loss', 'cohen_kappa']:
    print(f"{key:<20} "
          f"{results_100[key]:>10.4f} "
          f"{results_50[key]:>10.4f} "
          f"{results_masked[key]:>10.4f}")


# EXTRACT CNN EMBEDDINGS FOR GNN
class AllCellDataset(Dataset):
    """Loads all records in master_df for embedding extraction."""
    def __init__(self, df, use_100px=True):
        self.df        = df.reset_index(drop=True)
        self.path_col  = 'path_100' if use_100px else 'path_50'
        self.transform = get_transforms(use_100px)

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        img = Image.open(row[self.path_col]).convert('RGB')
        if row['augmented'] and row['rotation'] != 0:
            img = img.rotate(row['rotation'])
        return self.transform(img), row['label']


def extract_embeddings(model, master_df, use_100px=True):
    tag = '100px' if use_100px else '50px'
    print(f"\nExtracting embeddings — {tag}")

    all_ds     = AllCellDataset(master_df, use_100px)
    all_loader = DataLoader(all_ds, batch_size=BATCH_SIZE,
                            shuffle=False, num_workers=2, pin_memory=True)

    embeddings_list = []

    def hook_fn(module, input, output):
        embeddings_list.append(output.squeeze(-1).squeeze(-1).cpu())

    hook = model.avgpool.register_forward_hook(hook_fn)
    model.eval()
    with torch.no_grad():
        for i, (imgs, _) in enumerate(all_loader):
            _ = model(imgs.to(DEVICE))
            if i % 50 == 0:
                print(f"  Batch {i}/{len(all_loader)}")
    hook.remove()

    node_features = torch.cat(embeddings_list, dim=0)
    save_path     = OUT_DIR + f'gnn_node_features_{tag}.pt'
    torch.save(node_features, save_path)
    return node_features


# Extract embeddings for GNN (strategy 2 models only)
emb_100 = extract_embeddings(model_100, master_s2, use_100px=True)
emb_50  = extract_embeddings(model_50,  master_s2, use_100px=False)
```
